# WNBA Draft Fit Predictor — Data Collection

This notebook pulls and saves raw data for three sources:
1. NCAA career stats for 2019–2026 draft classes
2. WNBA rookie season stats for 2019–2025 draft classes
3. Current WNBA team stats and tendencies

Data is saved to /data/raw/ for cleaning in the next notebook.

```python
# Core libraries
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import time
import os

# Check all imports work
print("All imports successful!")
```

```python
import os

# List of all prospect file names
prospect_files = [
    ("Azzi Fudd", "azzi_fudd.csv"),
    ("Lauren Betts", "lauren_betts.csv"),
    ("Olivia Miles", "olivia_miles.csv"),
    ("Flau'jae Johnson", "flauajae_johnson.csv"),
    ("Gabriela Jaquez", "gabriela_jaquez.csv"),
    ("Kiki Rice", "kiki_rice.csv"),
    ("Gianna Kneepkens", "gianna_kneepkens.csv"),
    ("Ta'Niya Latson", "taniya_latson.csv"),
    ("Raven Johnson", "raven_johnson.csv"),
    ("Marta Suarez", "marta_suarez.csv"),
    ("Janiah Barker", "janiah_barker.csv"),
    ("Serah Williams", "serah_williams.csv"),
    ("Tonie Morgan", "tonie_morgan.csv"),
    ("Darianna Littlepage-Buggs", "darianna_littlepage_buggs.csv"),
]

all_prospects = []

for name, filename in prospect_files:
    filepath = f"../data/external/{filename}"
    df = pd.read_csv(filepath)
    df["player"] = name
    all_prospects.append(df)

prospects_df = pd.concat(all_prospects, ignore_index=True)
print(f"Total rows: {prospects_df.shape[0]}")
print(f"Total columns: {prospects_df.shape[1]}")
print(prospects_df["player"].value_counts())
```

```python
prospects_df.to_csv("../data/processed/prospects_2026_raw.csv", index=False)
print("Saved!")
```
